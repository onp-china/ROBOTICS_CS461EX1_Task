## 4. 提取指标与对比绘图

```python
def extract_metrics(log_file):
    if not os.path.exists(log_file):
        print(f"日志文件不存在: {log_file}")
        return pd.DataFrame()
    
    data = []
    with open(log_file, 'r') as f:
        for line in f:
            if not line.strip():
                continue
            try:
                row = json.loads(line)
                # 提取训练 loss 和测试成功率
                if 'epoch' in row:
                    entry = {'epoch': row['epoch']}
                    if 'train_loss' in row: entry['train_loss'] = row['train_loss']
                    if 'val_loss' in row: entry['val_loss'] = row['val_loss']
                    if 'test/mean_score' in row: entry['test_mean_score'] = row['test/mean_score']
                    data.append(entry)
            except json.JSONDecodeError:
                pass
    if not data:
        return pd.DataFrame()
    return pd.DataFrame(data).groupby('epoch').mean().reset_index()

def plot_comparison(task_name):
    # 根据我们设置的 seed 寻找日志
    baseline_log = f"mydiffusion/outputs/{task_name}_lowdim_abs/42/logs.json.txt"
    attnres_log = f"mydiffusion/outputs/{task_name}_lowdim_abs/43/logs.json.txt"
    
    df_base = extract_metrics(baseline_log)
    df_attn = extract_metrics(attnres_log)
    
    if df_base.empty and df_attn.empty:
        print(f"未找到 {task_name} 的有效日志数据。")
        return
        
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"Comparison: {task_name}", fontsize=14)
    
    # Plot Loss (优先显示 train_loss，如果没有则显示 val_loss)
    for df, label, color in [(df_base, 'Baseline', 'blue'), (df_attn, 'AttnRes', 'red')]:
        if not df.empty:
            if 'train_loss' in df:
                axes[0].plot(df['epoch'], df['train_loss'], label=f'{label} (Train)', color=color)
            elif 'val_loss' in df:
                axes[0].plot(df['epoch'], df['val_loss'], label=f'{label} (Val)', color=color, linestyle='--')
    
    axes[0].set_title('Loss Curve')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].grid(True)
    axes[0].legend()
    
    # Plot Success Rate
    for df, label, color in [(df_base, 'Baseline', 'blue'), (df_attn, 'AttnRes', 'red')]:
        if not df.empty and 'test_mean_score' in df:
            axes[1].plot(df['epoch'], df['test_mean_score'], label=label, color=color)
    
    axes[1].set_title('Test Success Rate')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Success Rate')
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(True)
    axes[1].legend()
    
    plt.tight_layout()
    plt.show()

print("可视化模块已就绪。使用示例: plot_comparison('mug_cleanup_d0')")
```
